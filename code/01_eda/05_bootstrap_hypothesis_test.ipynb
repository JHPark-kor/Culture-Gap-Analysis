- 조사연도×소득집단 층화 비모수 부트스트랩
- 범주별 Owen 격차설명량의 95% 신뢰구간과 단측 검정

```python
import pathlib
import sys
import pandas as pd

BASE_PATH = pathlib.Path.cwd().resolve()
if not (BASE_PATH / "data").exists():
    BASE_PATH = next(parent for parent in BASE_PATH.parents if (parent / "data").exists())
CODE_PATH = BASE_PATH / "code" / "01_eda"
if str(CODE_PATH) not in sys.path:
    sys.path.insert(0, str(CODE_PATH))
RAW_PATH = BASE_PATH / "data" / "raw"
PROCESSED_PATH = BASE_PATH / "data" / "processed" / "eda"
IMAGE_PATH = BASE_PATH / "output" / "image" / "readme"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)
IMAGE_PATH.mkdir(parents=True, exist_ok=True)
print("BASE_PATH:", BASE_PATH)
```

## 입력 데이터와 실행 설정

```python
from analysis.bootstrap import run_bootstrap
MODEL_FILE = PROCESSED_PATH / "03_model_analysis_frame.parquet"
MODEL_DIR = PROCESSED_PATH / "02_models"
OUTPUT_DIR = PROCESSED_PATH / "03_bootstrap"
BOOTSTRAP_REPS = 100
N_JOBS = 4
assert MODEL_FILE.exists() and (MODEL_DIR / "06_gap_category_owen.csv").exists()
model_frame = pd.read_parquet(MODEL_FILE)
```

## 층화 부트스트랩과 가설 검정

```python
outputs = run_bootstrap(model_frame, MODEL_DIR, OUTPUT_DIR, reps=BOOTSTRAP_REPS, jobs=N_JOBS)
ci = outputs["02_category_confidence_intervals.csv"]
test = outputs["03_opportunity_practical_contrast.csv"]
display(ci.round(3))
display(test.round(3))
```

## 결과 품질 점검

```python
replicates = outputs["01_bootstrap_replicates.csv"]
assert replicates["반복"].nunique() == BOOTSTRAP_REPS
assert len(replicates) == BOOTSTRAP_REPS * 4 * 7
```
