# 서울시 문화누리카드 분야별 이용 현황

## 목적
- 2025년 서울시 문화누리카드 이용건수를 분야별로 집계
- 도서·영화와 공연·음악·전시의 이용 비중 비교
- 자치구 합계와 전체 이용건수 검증

```python
import pathlib
import sys
import pandas as pd

BASE_PATH = pathlib.Path.cwd().resolve()
if not (BASE_PATH / "data").exists():
    BASE_PATH = next(parent for parent in BASE_PATH.parents if (parent / "data").exists())
CODE_PATH = BASE_PATH / "code" / "01_eda"
if str(CODE_PATH) not in sys.path:
    sys.path.insert(0, str(CODE_PATH))
RAW_PATH = BASE_PATH / "data" / "raw"
PROCESSED_PATH = BASE_PATH / "data" / "processed" / "eda"
IMAGE_PATH = BASE_PATH / "output" / "image" / "readme"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)
IMAGE_PATH.mkdir(parents=True, exist_ok=True)
print("BASE_PATH:", BASE_PATH)
```

## 입력 데이터와 결과 경로

```python
from analysis.descriptive import analyze_culture_nuri
SOURCE_FILE = RAW_PATH / "mnc_merchants" / "card_usage" / "mnc_seoul_usage_issuance_2021_2025.xlsx"
OUTPUT_DIR = PROCESSED_PATH / "01_descriptive"
assert SOURCE_FILE.exists(), SOURCE_FILE
```

## 분야별 이용건수와 비중 산출

```python
usage, summary = analyze_culture_nuri(SOURCE_FILE, OUTPUT_DIR)
display(usage.head(10))
display(usage.set_index("분야").loc[["도서", "영화", "공연", "음악", "전시"]].round(3))
```

## 결과 품질 점검

```python
assert summary["자치구수"] == 25
assert all(summary["검증"].values())
print("결과 저장:", OUTPUT_DIR.relative_to(BASE_PATH))
```
