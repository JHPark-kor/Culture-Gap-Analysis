# 소득집단별 기초예술 관람과 관람의향

## 목적
- 2023~2025년 전 연령 응답자 30,325명을 결합
- 최근 기초예술 관람률, 향후 관람의향률, 의향자 중 관람률 비교

```python
import pathlib
import sys
import pandas as pd

BASE_PATH = pathlib.Path.cwd().resolve()
if not (BASE_PATH / "data").exists():
    BASE_PATH = next(parent for parent in BASE_PATH.parents if (parent / "data").exists())
CODE_PATH = BASE_PATH / "code" / "01_eda"
if str(CODE_PATH) not in sys.path:
    sys.path.insert(0, str(CODE_PATH))
RAW_PATH = BASE_PATH / "data" / "raw"
PROCESSED_PATH = BASE_PATH / "data" / "processed" / "eda"
IMAGE_PATH = BASE_PATH / "output" / "image" / "readme"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)
IMAGE_PATH.mkdir(parents=True, exist_ok=True)
print("BASE_PATH:", BASE_PATH)
```

## 입력 데이터 설정

```python
from analysis.data import build_descriptive_frame, load_raw_waves
from analysis.descriptive import analyze_income_stages
SURVEY_PATH = RAW_PATH / "culture_survey" / "national_culture_arts_activity"
OUTPUT_DIR = PROCESSED_PATH / "01_descriptive"
for year in (2023, 2024, 2025):
    assert len(list(SURVEY_PATH.glob(f"{year}_*.csv"))) == 1
```

## 전 연령 분석자료 생성

```python
analysis_frame = build_descriptive_frame(load_raw_waves(SURVEY_PATH))
assert len(analysis_frame) == 30325
assert analysis_frame.isna().sum().sum() == 0
display(analysis_frame.groupby(["조사연도", "소득집단"]).size().rename("표본수").reset_index())
```

## 소득집단별 관람·의향 지표

```python
summary, states = analyze_income_stages(analysis_frame, OUTPUT_DIR)
display(summary.round(2))
display(states.round(2))
```

## 결과 품질 점검

```python
assert summary.set_index("소득집단")["표본수"].to_dict() == {"저소득 집단": 4366, "그 외 소득집단": 25959}
assert states.groupby("소득집단")["소득집단내_가중비율_퍼센트"].sum().round(8).eq(100).all()
```

- 기술적 비교이며 시간적 인과관계를 추정하지 않음
