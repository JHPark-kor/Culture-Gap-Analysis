# EDA·관람의향 분석 시각자료 생성

## 목적
- 1~5번 노트북 결과로 README용 그래프 생성
- 프로젝트 규칙에 따라 `output/image/readme/`에 저장

```python
import pathlib
import sys
import pandas as pd

BASE_PATH = pathlib.Path.cwd().resolve()
if not (BASE_PATH / "data").exists():
    BASE_PATH = next(parent for parent in BASE_PATH.parents if (parent / "data").exists())
CODE_PATH = BASE_PATH / "code" / "01_eda"
if str(CODE_PATH) not in sys.path:
    sys.path.insert(0, str(CODE_PATH))
RAW_PATH = BASE_PATH / "data" / "raw"
PROCESSED_PATH = BASE_PATH / "data" / "processed" / "eda"
IMAGE_PATH = BASE_PATH / "output" / "image" / "readme"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)
IMAGE_PATH.mkdir(parents=True, exist_ok=True)
print("BASE_PATH:", BASE_PATH)
```

## 입력 결과 확인

```python
from analysis.figures import build_all
DESCRIPTIVE_DIR = PROCESSED_PATH / "01_descriptive"
MODEL_DIR = PROCESSED_PATH / "02_models"
BOOTSTRAP_DIR = PROCESSED_PATH / "03_bootstrap"
required = [DESCRIPTIVE_DIR / "01_culture_nuri_field_usage.csv", DESCRIPTIVE_DIR / "03_income_stage_summary.csv", MODEL_DIR / "09_mean_intention_category_contribution.csv", MODEL_DIR / "10_mean_gap_category_contribution.csv", BOOTSTRAP_DIR / "03_opportunity_practical_contrast.csv"]
assert not [path for path in required if not path.exists()]
```

## 시각자료 생성

```python
build_all(DESCRIPTIVE_DIR, MODEL_DIR, BOOTSTRAP_DIR, IMAGE_PATH)
figures = sorted(IMAGE_PATH.glob("*.png"))
display(pd.DataFrame({"파일": [path.name for path in figures], "크기_바이트": [path.stat().st_size for path in figures]}))
```

## 결과 품질 점검

```python
expected = [f"0{number}_" for number in range(1, 9)]
generated = [path.name for path in IMAGE_PATH.glob("*.png")]
assert all(any(name.startswith(prefix) for name in generated) for prefix in expected)
```
